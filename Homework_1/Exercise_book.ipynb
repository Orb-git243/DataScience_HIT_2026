```python
# Get the top 10 rows of the dataset sorted by the "carat" column in descending order
# ascending=False means that the values will be sorted in descending order
diamonds.sort_values(by="carat", ascending=False).head(10)
```

```python
# Get the top 10 rows of the dataset sorted by the "carat" column in descending order and save the changes to the original dataset
diamonds.sort_values(by="carat", ascending=False, inplace=True)
diamonds.head(10)
```

```python
# Create a new column "price_per_carat" that is the price divided by the carat
diamonds["price_per_carat"] = diamonds["price"] / diamonds["carat"]
# Create a new column "is_expensive" that is True if the price is greater than 10000, False otherwise
diamonds["is_expensive"] = diamonds["price"] > 10000
diamonds.head(10)
```

```python
# Create a new column "rate" that is "expensive" if the price is greater than 10000, "cheap" otherwise
import numpy as np
diamonds["rate"] = np.where(diamonds["price"] > 10000, "expensive", "cheap")
diamonds.head()
```

```python
# Get the mean of the "price" column in the dataset
diamonds["price"].mean()
```
