```python
# Create a new column "price_per_carat" that is the price divided by the carat
diamonds["price_per_carat"] = diamonds["price"] / diamonds["carat"]
# Create a new column "is_expensive" that is True if the price is greater than 10000, False otherwise
diamonds["is_expensive"] = diamonds["price"] > 10000
diamonds.head(10)
```

```python
# Create a new column "rate" that is "expensive" if the price is greater than 10000, "cheap" otherwise
import numpy as np
diamonds["rate"] = np.where(diamonds["price"] > 10000, "expensive", "cheap")
diamonds.head()
```

```python
# calling the pandas library
import pandas as pd
import plotnine as pln

iris_df = pd.read_csv(r'C:\Users\orbar\Desktop\Study_materials_HIT\DataScience_HIT_2026\Practice_and_understanding_the_study\iris.csv', sep=',', index_col=0)
```

```python
print(iris_df.columns)
```

```python
pln.ggplot(
    iris_df, 
    # Define aesthetic mappings: x-axis, y-axis, and color grouping by species
    pln.aes(x='SepalLengthCm', y='SepalWidthCm', color='Species')
) + \
pln.geom_point() + \
pln.theme_bw() + \
pln.labs(
    title='Sepal Length vs Sepal Width', 
    x='Sepal Length (cm)', 
    y='Sepal Width (cm)'
)
```

```python
pln.ggplot(iris_df, pln.aes(x='SepalWidthCm')) + pln.geom_histogram() + pln.facet_wrap('Species')
```

```python
pln.ggplot(iris_df, pln.aes(x='SepalWidthCm')) + pln.geom_histogram() + pln.facet_wrap('Species', nrow = 3)
```

```python
setosa_df = iris_df.loc[iris_df['Species'] == 'Iris-setosa']
versicolor_df = iris_df.loc[iris_df['Species'] == 'Iris-versicolor']
virginica_df = iris_df.loc[iris_df['Species'] == 'Iris-virginica']

pln.ggplot() + \
    pln.geom_histogram(setosa_df, pln.aes(x='SepalWidthCm'), fill='blue', alpha=0.5) + \
    pln.geom_histogram(versicolor_df, pln.aes(x='SepalWidthCm'), binwidth=0.2, fill='green', alpha=0.5)
```

```python
pln.ggplot(iris_df, pln.aes(x='Species', y='SepalWidthCm', fill = 'Species')) + pln.geom_boxplot() + pln.geom_jitter(width=0.1)
```
